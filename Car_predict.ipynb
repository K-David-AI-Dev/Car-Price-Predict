```python
import pandas as pd

# 1. Load the dataset
df = pd.read_csv('used_car_dataset.csv')

# 2. List column names
print("Dataset columns:")
print(df.columns.tolist())

# 3. Display the first 5 rows
print("\nFirst 5 rows of the dataset:")
print(df.head())

# 4. Basic information (data types, missing values)
print("\nDataset information:")
print(df.info())
```

```python
# 1. Dropping unnecessary columns
df = df.drop(['PostedDate', 'AdditionInfo'], axis=1)

# 2. Verification
print("Columns after dropping 'PostedDate' and 'AdditionInfo':")
print(df.columns.tolist())

# 3. Displaying the first 5 rows to confirm
print("\nFirst 5 rows after column removal:")
print(df.head())
```

```python
import pandas as pd

# 1. Cleaning 'AskPrice': removing '₹', commas and leading/trailing spaces
df['AskPrice'] = df['AskPrice'].str.replace('₹', '', regex=False)
df['AskPrice'] = df['AskPrice'].str.replace(',', '', regex=False).str.strip()

# 2. Converting to numeric (float)
df['AskPrice'] = pd.to_numeric(df['AskPrice'], errors='coerce')

# 3. Converting from INR to EUR (Rate: 1 INR = 0.011 EUR)
exchange_rate = 0.011
df['Price'] = df['AskPrice'] * exchange_rate

# 4. Dropping the old 'AskPrice' column
df = df.drop(columns=['AskPrice'])

# 5. Display results
print("Price column successfully cleaned and converted to EUR.")
print("\nFirst 5 rows of the new Price column:")
print(df[['Price']].head())

print("\nUpdated Dataset Information:")
print(df.info())
```

```python
import pandas as pd
import numpy as np

# 1. Cleaning 'kmDriven': removing 'km', commas and spaces
df['kmDriven'] = df['kmDriven'].str.replace('km', '', regex=False)
df['kmDriven'] = df['kmDriven'].str.replace(',', '', regex=False).str.strip()

# 2. Converting to numeric (float)
df['kmDriven'] = pd.to_numeric(df['kmDriven'], errors='coerce')

# 3. Filling missing values with the mean of the column
km_mean = df['kmDriven'].mean()
df['kmDriven'] = df['kmDriven'].fillna(km_mean)

# 4. Display results
print(f"kmDriven column successfully cleaned.")
print(f"Missing values filled with mean: {km_mean:.2f}")

print("\nFirst 5 rows of the cleaned kmDriven column:")
print(df[['kmDriven']].head())

print("\nUpdated Dataset Information:")
print(df.info())
```

```python
import pandas as pd

# 1. Check unique values in Transmission
print("Unique values in Transmission before encoding:")
print(df['Transmission'].unique())

# 2. Perform One-Hot Encoding
# This will create new columns like Transmission_Manual, Transmission_Automatic
df = pd.get_dummies(df, columns=['Transmission'], prefix='Transmission')

# 3. Convert boolean values (True/False) to integers (1/0)
for col in df.columns:
    if 'Transmission_' in col:
        df[col] = df[col].astype(int)

# 4. Display results
print("\nTransmission successfully encoded.")
print("Current columns in the dataset:")
print(df.columns.tolist())

print("\nFirst 5 rows of the encoded Transmission columns:")
transmission_cols = [col for col in df.columns if 'Transmission_' in col]
print(df[transmission_cols].head())
```

```python
import pandas as pd

# 1. Check unique values in Owner
print("Unique values in Owner before encoding:")
print(df['Owner'].unique())

# 2. Perform One-Hot Encoding for Owner
df = pd.get_dummies(df, columns=['Owner'], prefix='Owner')

# 3. Convert boolean values to integers (1/0)
for col in df.columns:
    if 'Owner_' in col:
        df[col] = df[col].astype(int)

# 4. Display results
print("\nOwner column successfully encoded.")
print("Current columns in the dataset:")
print(df.columns.tolist())

# 5. Show the first 5 rows of the new Owner columns
owner_cols = [col for col in df.columns if 'Owner_' in col]
print("\nFirst 5 rows of the encoded Owner columns:")
print(df[owner_cols].head())
```

```python
import pandas as pd

# 1. Check unique values in FuelType
print("Unique values in FuelType before encoding:")
print(df['FuelType'].unique())

# 2. Perform One-Hot Encoding for FuelType
df = pd.get_dummies(df, columns=['FuelType'], prefix='FuelType')

# 3. Convert boolean values to integers (1/0)
for col in df.columns:
    if 'FuelType_' in col:
        df[col] = df[col].astype(int)

# 4. Display results
print("\nFuelType successfully encoded.")
print("Current columns in the dataset:")
print(df.columns.tolist())

# 5. Show the first 5 rows of the new FuelType columns
fuel_cols = [col for col in df.columns if 'FuelType_' in col]
print("\nFirst 5 rows of the encoded FuelType columns:")
print(df[fuel_cols].head())
```

```python
# 1. Cleaning the Brand column (stripping spaces and unifying to Title Case)
df['Brand'] = df['Brand'].str.strip().str.title()

# 2. Counting occurrences for each brand
brand_counts = df['Brand'].value_counts()

# 3. Displaying results
print("Brand distribution in the dataset:")
print(brand_counts)

print(f"\nTotal number of unique brands after cleaning: {len(brand_counts)}")
```

```python
# 1. Count occurrences again
brand_counts = df['Brand'].value_counts()

# 2. Identify brands with at least 10 occurrences
brands_to_keep = brand_counts[brand_counts >= 10].index

# 3. Filter the DataFrame
df = df[df['Brand'].isin(brands_to_keep)]

# 4. Display the results
print(f"Filtering complete. Brands with less than 10 occurrences have been removed.")
print(f"Remaining unique brands: {df['Brand'].nunique()}")
print(f"Total rows remaining in dataset: {len(df)}")

print("\nTop 10 brands after filtering:")
print(df['Brand'].value_counts().head(10))
```

```python
import numpy as np

# 1. CALCULATE MEAN PRICE PER BRAND (Target Encoding Base)
# We group by Brand and calculate the average Price
brand_means = df.groupby('Brand')['Price'].mean()

# 2. SAVE MAPPING DICTIONARY (Crucial Step for Future Predictions)
# We save this mapping to a dictionary before dropping the original Brand column
df_brand_mapping = brand_means.to_dict()

# 3. APPLY TARGET ENCODING WITH 0.01 NOISE
# Setting seed for reproducibility
seed_value = 42
np.random.seed(seed_value)

# Calculate noise (1% of Price standard deviation)
price_std = df['Price'].std()
noise = 0.01 * price_std * np.random.randn(len(df))

# Map the mean values to the rows and add Gaussian noise
df['Brand_encoded'] = df['Brand'].map(df_brand_mapping) + noise

# 4. REMOVE ORIGINAL CATEGORICAL COLUMN
# Now it is safe to drop 'Brand' as the relationship is stored in df_brand_mapping
df = df.drop(columns=['Brand'])

# 5. VERIFICATION
print("Brand target encoding completed successfully.")
print("The mapping dictionary 'df_brand_mapping' is stored in memory.")

# Print sample of the mapping to confirm
print("\nSample of the saved Brand mapping:")
for brand, value in list(df_brand_mapping.items())[:5]:
    print(f"{brand}: {value:.2f} EUR")
```

```python
# 1. Cleaning the 'model' column (stripping spaces and unifying to Title Case)
df['model'] = df['model'].str.strip().str.title()

# 2. Counting occurrences for each model
model_counts = df['model'].value_counts()

# 3. Displaying results
print("Model distribution in the dataset (Top 20):")
print(model_counts.head(20))

print(f"\nTotal number of unique models after cleaning: {len(model_counts)}")
```

```python
# 1. Count occurrences for each model
model_counts = df['model'].value_counts()

# 2. Identify models with at least 20 occurrences
models_to_keep = model_counts[model_counts >= 20].index

# 3. Filter the DataFrame to keep only those models
df = df[df['model'].isin(models_to_keep)]

# 4. Display the results
print(f"Filtering complete. Models with less than 20 occurrences have been removed.")
print(f"Remaining unique models: {df['model'].nunique()}")
print(f"Total rows remaining in dataset: {len(df)}")

print("\nTop 10 models after filtering:")
print(df['model'].value_counts().head(10))
```

```python
import numpy as np

# 1. CALCULATE MEAN PRICE PER MODEL (Target Encoding Base)
# We group by the 'model' column and calculate the average price for each
model_means = df.groupby('model')['Price'].mean()

# 2. SAVE MAPPING DICTIONARY (Essential for Future Predictions)
# We store the relationship between model names and mean prices
df_model_mapping = model_means.to_dict()

# 3. APPLY TARGET ENCODING WITH 0.01 NOISE
# Setting seed for reproducibility
seed_value = 42
np.random.seed(seed_value)

# Calculate Gaussian noise based on 1% of the Price standard deviation
price_std = df['Price'].std()
noise_model = 0.01 * price_std * np.random.randn(len(df))

# Create the encoded column by mapping the names to means and adding noise
df['model_encoded'] = df['model'].map(df_model_mapping) + noise_model

# 4. REMOVE ORIGINAL CATEGORICAL COLUMN
# With the mapping safely stored in df_model_mapping, we can drop the text column
df = df.drop(columns=['model'])

# 5. VERIFICATION
print("Model target encoding completed successfully.")
print("The mapping dictionary 'df_model_mapping' is ready for use.")

# Print sample of the mapping to confirm data integrity
print("\nSample of the saved Model mapping:")
for car_model, value in list(df_model_mapping.items())[:5]:
    print(f"{car_model}: {value:.2f} EUR")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error, r2_score

# 1. Separate features (X) and target variable (y)
# We predict 'Price', so that is our y. Everything else is X.
X = df.drop('Price', axis=1)
y = df['Price']

# 2. Split the data (80% Train, 20% Test)
# random_state ensures we get the same split every time we run it
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

# 3. Initialize and Train the Decision Tree Regressor
# We use a regressor because we are predicting a continuous number (Price)
dt_model = DecisionTreeRegressor(random_state=42)
dt_model.fit(X_train, y_train)

# 4. Make predictions on the test set
y_pred = dt_model.predict(X_test)

# 5. Evaluate the model
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Model Training Completed.")
print(f"Mean Absolute Error (MAE): {mae:.2f} EUR")
print(f"R-squared Score (Accuracy): {r2:.4f}")

# 6. Check feature importance
importance = pd.DataFrame({'Feature': X.columns, 'Importance': dt_model.feature_importances_})
print("\nFeature Importances:")
print(importance.sort_values(by='Importance', ascending=False))
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, r2_score

# 1. Initialize the Random Forest Regressor
# n_estimators=100 means we are using 100 different decision trees
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)

# 2. Train the model
rf_model.fit(X_train, y_train)

# 3. Make predictions on the test set
y_pred_rf = rf_model.predict(X_test)

# 4. Evaluate the Random Forest model
mae_rf = mean_absolute_error(y_test, y_pred_rf)
r2_rf = r2_score(y_test, y_pred_rf)

print("Random Forest Training Completed.")
print(f"Mean Absolute Error (MAE): {mae_rf:.2f} EUR")
print(f"R-squared Score (Accuracy): {r2_rf:.4f}")

# 5. Compare Feature Importance
rf_importance = pd.DataFrame({'Feature': X.columns, 'Importance': rf_model.feature_importances_})
print("\nRandom Forest Feature Importances (Top 10):")
print(rf_importance.sort_values(by='Importance', ascending=False).head(10))
```

```python
import xgboost as xg
from sklearn.metrics import mean_absolute_error, r2_score

# 1. Initialize the XGBoost Regressor
# n_estimators=150 as requested
xgb_model = xg.XGBRegressor(n_estimators=150, objective='reg:squarederror', random_state=42)

# 2. Train the model
xgb_model.fit(X_train, y_train)

# 3. Make predictions
y_pred_xgb = xgb_model.predict(X_test)

# 4. Evaluate the XGBoost model
mae_xgb = mean_absolute_error(y_test, y_pred_xgb)
r2_xgb = r2_score(y_test, y_pred_xgb)

print("XGBoost Training Completed (150 estimators).")
print(f"Mean Absolute Error (MAE): {mae_xgb:.2f} EUR")
print(f"R-squared Score (Accuracy): {r2_xgb:.4f}")

# 5. Feature Importance for XGBoost
xgb_importance = pd.DataFrame({'Feature': X.columns, 'Importance': xgb_model.feature_importances_})
print("\nXGBoost Feature Importances (Top 10):")
print(xgb_importance.sort_values(by='Importance', ascending=False).head(10))
```

```python
import pandas as pd
import numpy as np

# 1. Define the input car data with the specified mileage
input_brand = "Skoda"
input_model = "Fabia"
input_year = 2011
input_age = 2026 - input_year
input_km = 200000 # Your updated mileage
input_transmission = "Manual"
input_owner = "second"
input_fuel = "Diesel"

# 2. Convert categorical data using our saved mappings
# If a brand/model is not in the mapping, we use the global average
brand_val = df_brand_mapping.get(input_brand, np.mean(list(df_brand_mapping.values())))
model_val = df_model_mapping.get(input_model, np.mean(list(df_model_mapping.values())))

# 3. Prepare the input features (matching the X_train structure)
input_data = {col: 0 for col in X.columns}

# Fill numerical features
input_data['Year'] = input_year
input_data['Age'] = input_age
input_data['kmDriven'] = input_km
input_data['Brand_encoded'] = brand_val
input_data['model_encoded'] = model_val

# Fill One-Hot encoded features
if f'Transmission_{input_transmission}' in input_data:
    input_data[f'Transmission_{input_transmission}'] = 1
if f'Owner_{input_owner}' in input_data:
    input_data[f'Owner_{input_owner}'] = 1
if f'FuelType_{input_fuel}' in input_data:
    input_data[f'FuelType_{input_fuel}'] = 1

# Convert to DataFrame
input_df = pd.DataFrame([input_data])

# 4. Get updated predictions from all models
pred_dt = dt_model.predict(input_df)[0]
pred_rf = rf_model.predict(input_df)[0]
pred_xgb = xgb_model.predict(input_df)[0]

# 5. Display the final results
print(f"--- Updated Prediction for: {input_year} {input_brand} {input_model} ---")
print(f"Details: {input_fuel}, {input_transmission}, {input_owner} owner, {input_km} km\n")
print(f"Decision Tree Prediction:  {pred_dt:.2f} EUR")
print(f"Random Forest Prediction:  {pred_rf:.2f} EUR")
print(f"XGBoost Prediction:        {pred_xgb:.2f} EUR")
```

```python
import joblib
from google.colab import files
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor

# 1. INTEGRATED PIPELINE PREPARATION
# Since we already performed Target Encoding and splitting,
# we wrap our best model into a Pipeline object.
rf_pipeline = Pipeline(steps=[
    ('regressor', RandomForestRegressor(n_estimators=100, random_state=42))
])

# 2. TRAINING THE PIPELINE
# We fit the pipeline using the data we prepared (X_train, y_train)
rf_pipeline.fit(X_train, y_train)

# 3. BUNDLING ALL NECESSARY DATA
# We include the model, the mappings, and the mean mileage for full portability
export_data = {
    'pipeline': rf_pipeline,
    'brand_mapping': df_brand_mapping,
    'model_mapping': df_model_mapping,
    'km_mean': km_mean,
    'feature_names': X.columns.tolist()
}

# 4. SAVING AND DOWNLOADING
filename = 'car_price_full_pipeline.pkl'
joblib.dump(export_data, filename)

print(f"Status: Pipeline and Mappings bundled into '{filename}'.")
print("Action: Triggering browser download...")
print("Pipeline successfully created and saved as 'car_price_predictor_pipeline.pkl'.")

# This command will prompt your browser to save the file
files.download(filename)
```
